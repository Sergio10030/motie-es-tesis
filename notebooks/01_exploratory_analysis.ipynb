# Análisis exploratorio

Exploración inicial de los datos de reportes escolares.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = next((path for path in [Path.cwd(), *Path.cwd().parents] if (path / "config/config.yaml").exists()), Path.cwd())
DATA_PATH = ROOT / "data/processed/corpus_processed.csv"
if not DATA_PATH.exists():
    raise FileNotFoundError("No existe el corpus procesado. Ejecuta primero: python -m experiments.run_prepare_data")

df = pd.read_csv(DATA_PATH, encoding="utf-8")
print(f"Registros: {len(df):,} | Columnas: {len(df.columns)}")
display(df.head())

quality = pd.DataFrame({
    "nulos": df.isna().sum(),
    "porcentaje_nulos": (100 * df.isna().mean()).round(2),
    "valores_unicos": df.nunique(dropna=True),
})
display(quality.sort_values("nulos", ascending=False).head(12))

for column in ["tipo_funcional", "prioridad"]:
    if column in df:
        counts = df[column].fillna("sin etiqueta").value_counts()
        display(counts.rename_axis(column).to_frame("reportes"))

if "n_tokens" in df:
    display(df["n_tokens"].describe().to_frame("tokens por reporte"))
```
