# Brazo B: fine-tuning

Entrenamiento y evaluación del modelo ajustado.

```python
import os
from pathlib import Path

import pandas as pd
import yaml
from IPython.display import display

ROOT = next((path for path in [Path.cwd(), *Path.cwd().parents] if (path / "config/config.yaml").exists()), Path.cwd())
os.chdir(ROOT)
config = yaml.safe_load((ROOT / "config/config.yaml").read_text(encoding="utf-8"))
processed = ROOT / config["paths"]["data_processed"]
train_path = processed / "train.csv"
val_path = processed / "val.csv"
test_path = processed / "test.csv"

if not all(path.exists() for path in [train_path, val_path, test_path]):
    raise FileNotFoundError("Faltan particiones. Ejecuta antes el notebook exploratorio o experiments/run_prepare_data.py.")
display(pd.DataFrame({"partición": ["train", "val", "test"], "registros": [len(pd.read_csv(path)) for path in [train_path, val_path, test_path]]}))
print(f"Modelo configurado: {config['arm_b']['default_model']}")
print("El fine-tuning descargará pesos y puede requerir GPU y bastante memoria.")

RUN_FINETUNING = False
if RUN_FINETUNING:
    from experiments.run_arm_b import main as run_arm_b
    run_arm_b(str(ROOT / "config/config.yaml"))
else:
    print("Fine-tuning omitido. Cambia RUN_FINETUNING a True para iniciarlo explícitamente.")
```
