```python
import os
from pathlib import Path

import pandas as pd
import yaml
from IPython.display import display

ROOT = next((path for path in [Path.cwd(), *Path.cwd().parents] if (path / "config/config.yaml").exists()), Path.cwd())
os.chdir(ROOT)
config = yaml.safe_load((ROOT / "config/config.yaml").read_text(encoding="utf-8"))
processed = ROOT / config["paths"]["data_processed"]
train_path = processed / "train.csv"
if not train_path.exists():
    raise FileNotFoundError("Faltan datos procesados. Ejecuta experiments/run_prepare_data.py primero.")
train = pd.read_csv(train_path, encoding="utf-8")

display(train[["texto", "tipo_funcional", "prioridad"]].head())
provider = next(item for item in config["arm_c"]["providers"] if item["name"] == config["arm_c"]["default_provider"])
print(f"Proveedor/modelo configurado: {provider['name']} / {provider['model']}")
print(f"API key configurada: {bool(os.environ.get(provider['api_key_env']))}")
print("Sin API key, el cliente usa respuestas simuladas; esas respuestas no son una evaluación del modelo real.")

RUN_LLM_EXPERIMENT = False
if RUN_LLM_EXPERIMENT:
    from experiments.run_arm_c import main as run_arm_c
    run_arm_c(str(ROOT / "config/config.yaml"))
else:
    print("Experimento omitido. Cambia RUN_LLM_EXPERIMENT a True para iniciarlo explícitamente.")
```
