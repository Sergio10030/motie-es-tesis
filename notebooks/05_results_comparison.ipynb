# Comparación de resultados

Comparación de métricas entre los brazos experimentales.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import display

ROOT = next((path for path in [Path.cwd(), *Path.cwd().parents] if (path / "config/config.yaml").exists()), Path.cwd())
metrics_dir = ROOT / "results/metrics"
rows = []
for arm in ["arm_a", "arm_b", "arm_c"]:
    path = metrics_dir / f"{arm}_metrics.json"
    if not path.exists():
        continue
    metrics = json.loads(path.read_text(encoding="utf-8"))
    rows.append({"brazo": arm.upper(), "modo": metrics.get("execution_mode", "baseline"), **{key: metrics.get(key) for key in ["accuracy", "f1_macro", "f1_weighted", "f2_macro", "accuracy_prioridad", "f1_macro_prioridad", "f1_weighted_prioridad"]}})

if rows:
    comparison = pd.DataFrame(rows).sort_values("f1_macro", ascending=False, na_position="last")
    display(comparison)
else:
    print("No hay métricas guardadas. Ejecuta primero los experimentos que quieras comparar.")

print("Compara solo resultados calculados sobre las mismas particiones y etiquetas; el dataset disponible usa etiquetas aproximadas derivadas de Tipo/Estado.")
```
